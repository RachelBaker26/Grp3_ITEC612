```python
# Repo Download

#basic imports
import os
import zipfile
import requests

#filenames
github_repo_zip_url = "https://github.com/RachelBaker26/Grp3_ITEC612/archive/refs/heads/main.zip"
embeddings_file = "face_embeddings.npz"

#load
print("Downloading dataset...")

r = requests.get(github_repo_zip_url, stream=True)
with open("repo.zip", "wb") as f:
  for chunk in r.iter_content(1024):
    f.write(chunk)

with zipfile.ZipFile("repo.zip", "r") as zip_ref:
  zip_ref.extractall("repo")
print("Files Downloaded")

```

```python
# for data/filesystem management
class FileSystem:
  def __init__(self):
    self.test = "/content/repo/Grp3_ITEC612-main/preprocess_aligned/Facial_Recognition_Test"
    self.train = "/content/repo/Grp3_ITEC612-main/preprocess_aligned/Facial_Recognition_Train"
    self.val = "/content/repo/Grp3_ITEC612-main/preprocess_aligned/Facial_Recognition_Val"
    self.peep_tuple = ("A_Daddario", "A_Samberg", "B_Eilish", "B_Pitt", "C_Boseman", "C_Theron", "E_Moss", "H_Jackman", "J_Travolta", "T_Cruise")
  def getSubset(self, variant):
    if variant.lower() == "test":
      return self.test
    if variant.lower() == "train":
      return self.train
    if variant.lower() == "val":
      return self.val
  def makeLink(self, peep, variant, img=""):
    #if this function is working correctly, you import a peep from peep tuple, and make the subset(tr,tst,val) link for that peep. so like Test/B_Pitt, once you make a link like that you can look over all imgs in that directory.
    bulkLink = self.getSubset(variant)
    if img == "":
      finishedLink = bulkLink + "/" + peep
    else:
      finishedLink = bulkLink + "/" + peep + "/" + img
    return finishedLink
  def makeEpoch(self, variant):
    #this function will try to return the path of all images of a subset(tr,tst,val), for all peeps once; This is mostly good for training, but maybe you find other uses for it.
    fileListOut = []
    for peep in self.peep_tuple:
      peepDir = self.makeLink(peep, variant)
      tempList=os.listdir(peepDir)
      for path in tempList:
        fileListOut.append(self.makeLink(peep,variant,path))
    return fileListOut

dbgFS=FileSystem()
print(dbgFS.makeEpoch("val"))
print(dbgFS.makeEpoch("train"))
print(dbgFS.makeEpoch("test"))
```
